```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 2D Line Plot

```python
# study hours of a student in a week

hours = [2, 3, 4, 1, 5, 7, 3]
days = [1, 2, 3, 4, 5, 6, 7]

plt.plot(days, hours)
```

```python
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Study hours of a student in a week')

plt.plot(days, hours)
```

```python
# showing study hours of multiple students in a week
hours_1 = [2, 3, 4, 1, 5, 7, 3]
hours_2 = [1, 4, 3, 5, 7, 2, 1]
days = [1, 2, 3, 4, 5, 6, 7]

plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Study hours of a student in a week')

plt.plot(days, hours_1)
plt.plot(days, hours_2)
```

## 2d line plot from pandas data

```python
# load file

df = pd.read_csv('enrollment_data.csv')
df
```

```python
plt.xlabel('Years')
plt.ylabel('Count of Enrolment')
plt.title('Students enrollmnt over 8 years')

plt.plot(df['Year'], df['Programming'])
```

```python
plt.xlabel('Years')
plt.ylabel('Count of Enrolment')
plt.title('Students enrollmnt over 8 years')

plt.plot(df['Year'], df['Programming'])
plt.plot(df['Year'], df['Digital Marketing'])
```

```python
# adding lebels using plt.Legend()
plt.xlabel('Years')
plt.ylabel('Count of Enrolment')
plt.title('Students enrollmnt over 8 years')

plt.plot(df['Year'], df['Programming'],label= 'Progeamming')
plt.plot(df['Year'], df['Digital Marketing'],label= 'Digital Marketing')

# plt.legend(loc='lower left')
plt.legend()
```

```python
# adding some visualization
#google search = plt line style

plt.xlabel('Years')
plt.ylabel('Count of Enrolment')
plt.title('Students enrollmnt over 8 years')

plt.plot(df['Year'], df['Programming'],label= 'Progeamming', color= '#26ced4', linewidth= 3, marker = 'o') #, linestyle = 'dashed'
plt.plot(df['Year'], df['Digital Marketing'],label= 'Digital Marketing', color= '#1d76cf', linewidth= 2, linestyle = 'dashdot')

plt.legend()
plt.grid()
plt.show()
```

## Scatter Plot

```python
students = pd.read_csv('student_IQdata.csv')

students
```

```python
#Relation between Study Hours and IQ Score -> positive co-relation
plt.xlabel('Study Hours')
plt.ylabel('IQ Score')
plt.title('Relation between Study Hours and IQ Score')

plt.scatter(students['Study_Hour'], students['IQ_Score'])
```

```python
#Relation between Shoe_Size and IQ Score -> no corelation
plt.xlabel('Shoe_Size')
plt.ylabel('IQ Score')
plt.title('Relation between Shoe_Size and IQ Score')

plt.scatter(students['Shoe_Size'], students['IQ_Score'])
```

```python
#Relation between Chilling_Hours and IQ Score -> negetive crelation
plt.xlabel('Chilling_Hours')
plt.ylabel('IQ Score')
plt.title('Relation between Shoe_Size and IQ Score')

plt.scatter(students['Chilling_Hours'], students['IQ_Score'])
```
